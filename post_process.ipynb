```python
import json
```

```python
# Load the JSON file
with open('./jsons/pilot3_alex_raw.json', 'r') as infile:
    data = json.load(infile)

unique_data = []
seen = set()

for entry in data:
    # Convert each dictionary to a tuple of key-value pairs for hashing
    entry_tuple = tuple((k, json.dumps(v, sort_keys=True)) for k, v in entry.items())
    if entry_tuple not in seen:
        seen.add(entry_tuple)
        unique_data.append(entry)

# Save the unique data to a new JSON file
with open('./jsons/pilot3_alex_clean.json', 'w') as outfile:
    json.dump(unique_data, outfile, indent=4)
```
